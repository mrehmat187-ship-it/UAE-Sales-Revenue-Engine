```python
import pandas as pd


sales_data = {
    'Order_ID': tuple((101, 102, 103, 104, 105, 106)),
    'City': ['Dubai', 'RAK', 'Dubai', 'Abu Dhabi', 'RAK', 'Dubai'],
    'Category': ['Electronics', 'Fashion', 'Electronics', 'Beauty', 'Electronics', 'Fashion'],
    'Revenue': [1200, 450, 2500, 300, 1800, 600]
}

df = pd.DataFrame(sales_data)


city_sales = df.groupby('City') ['Revenue'].sum()
print("--- Total Revenue by City ---")
print(city_sales)


top_category = df.groupby('Category')['Revenue'].sum()
print("\n--- Total Revenue by Category ---")
print(top_category)
```
